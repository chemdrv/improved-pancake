```python
import numpy as np

from astropy import units as u
from astropy import coordinates, time

from matplotlib import pyplot as plt

from tqdm.auto import tqdm
```

```python
equator = coordinates.EarthLocation.from_geodetic(0,0,0)
balt = coordinates.EarthLocation.of_address('3736 Tudor Arms Ave, Baltimore, MD 21211')
```

```python
t = time.Time('2025-3-21')
yr = t - np.linspace(0,1,2048)*u.yr
```

```python
sun = coordinates.get_sun(yr)
```

```python
plt.plot(yr.jd, sun.dec)
```

```python
sunaaeq = sun.transform_to(coordinates.AltAz(obstime=yr, location=equator))
plt.plot(yr.jd, sunaaeq.alt)
```

# Calculate local altitudes

```python
prevyrdays = t + np.arange(365)*u.day - 1*u.year
loc = equator

sunalts =  []
for d in tqdm(prevyrdays):
    daysamples = d + np.linspace(-12, 12, 2048)*u.hr

    aa = coordinates.AltAz(obstime=daysamples, location=loc)

    sunaa = coordinates.get_sun(daysamples).transform_to(aa)
    sunalts.append(sunaa.alt)

sunalts = u.Quantity(sunalts)
```

```python
plt.plot(sunalts.T);
```

```python
plt.scatter(days_from_jan1, np.mean(sunalts>0,axis=1), s=1)
plt.xlabel('days from Jan 1, 2025')
plt.ylabel('fraction of day with sun above horizon')

plt.title('Equator');
```

# Balt

```python
prevyrdays = t + np.arange(365)*u.day - 1*u.year
loc = balt

sunalts =  []
for d in tqdm(prevyrdays):
    daysamples = d + np.linspace(-12, 12, 2048)*u.hr

    aa = coordinates.AltAz(obstime=daysamples, location=loc)

    sunaa = coordinates.get_sun(daysamples).transform_to(aa)
    sunalts.append(sunaa.alt)

sunalts = u.Quantity(sunalts)
```

```python
plt.plot(sunalts.T);
```

```python
days_from_jan1 = prevyrdays.jd - time.Time('2025-1-1').jd
```

```python
plt.scatter(days_from_jan1, np.mean(sunalts>0,axis=1), s=1)
plt.xlabel('days from Jan 1, 2025')
plt.ylabel('fraction of day with sun above horizon')

plt.title('Baltimore');

plt.savefig('baltfrac.svg')
plt.savefig('baltfrac.png')
```
